```python
import pandas as pd
```

```python
july_test = pd.read_parquet('bookings_test_july.parquet')
may_jun_train = pd.read_parquet('bookings_train_may_june.parquet')
```

```python
#changing to valid datetime not objects
date_cols = [
    "booking_timestamp",
    "cancellation_date",
    "check_in_date",
    "check_out_date",
    "member_enrollment_date",
]

for df in (july_test, may_jun_train):
    for col in date_cols:
        df[col] = pd.to_datetime(df[col], errors="coerce")

#create variable that shows account age

df["account_age_days_at_booking"] = (
    df["booking_timestamp"] - df["member_enrollment_date"]
).dt.days
```
