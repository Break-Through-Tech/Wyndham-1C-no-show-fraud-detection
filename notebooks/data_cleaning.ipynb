```python
import pandas as pd
```

This file cleans the parquet files and adds variables needed for the models:
* Changes the dates to datetime types, instead of objects
* Adds an account age variable, calculates in days how long an account was created before its booking

After making changes to this file or doing further cleaning, please run the code at the end to update the changes in the parquet file and push to the github!

```python
july_test = pd.read_parquet('bookings_test_july.parquet')
may_jun_train = pd.read_parquet('bookings_train_may_june.parquet')
```

```python
#changing to valid datetime not objects
date_cols = [
    "booking_timestamp",
    "cancellation_date",
    "check_in_date",
    "check_out_date",
    "member_enrollment_date",
]
for df in (july_test, may_jun_train):
    for c in date_cols:
        df[c] = pd.to_datetime(df[c], errors="coerce", utc=True).dt.tz_localize(None)
    df = df.sort_values("booking_timestamp").reset_index(drop=True)

# for df in (july_test, may_jun_train):
#     for col in date_cols:
#         df[col] = pd.to_datetime(df[col], errors="coerce")

#create variable that shows account age
for df in (july_test, may_jun_train):
    df["account_age"] = (
        df["booking_timestamp"] - df["member_enrollment_date"]
    ).dt.days
```
